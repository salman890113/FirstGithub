# 3 Dictionaries
1-An ordered collection of elements
2-key and value
3-curly brackets/braces {}
4-change the values

```python
# food and their prices
food = {"pizza": 200 ,"burger" :140 ,"fries" :50 ,"zinger":250}
food
```

```python
type(food)
```

```python
# extracting the data 
food1=food.keys()
food1
```

```python
food1=food.values()
food1
```

```python
# adding new elements
food["raita"]=20
food
```

```python
# updating values
food["raita"]=14
food
```
